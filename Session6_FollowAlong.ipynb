## Section 3 · Filtering rows

Same idea as a NumPy boolean mask, but the mask is built from a **column**. `df[df["col"] > value]` keeps only the rows where the condition is True.

### 3.1 · One condition

```python
high = df[df["employment_rate"] > 92]

print(high.shape)

high.head()
```

### 3.2 · Two conditions — use `&` (and) / `|` (or), each side in ( )

```python
recent_high = df[(df["employment_rate"] > 92) & (df["year"] >= 2027)]
recent_high[["region", "year", "employment_rate"]].head()
```

## Section 4 · Cleaning — missing values, duplicates & messy text

Real data is messy. This file has three problems on purpose: **missing budgets**, a **duplicate row**, and a **mis-typed region label**. We fix each in code.

### 4.1 · Missing values

```python
df.isna().sum()
```

```python
print(df.isna().sum())                       # missing count per column

mean_budget = df["budget_million"].mean()


# print("mean budget:", round(mean_budget, 2))
```

```python
df_clean = df.copy()

df_clean["budget_million"] = df_clean["budget_million"].fillna(mean_budget)

print(df_clean["budget_million"].isna().sum())   # 0 -> all filled

print(df_clean)
```

### 4.2 · Duplicate rows

```python
print("duplicates:", df_clean.duplicated().sum())   # 1
df_clean = df_clean.drop_duplicates()
print("after drop:", df_clean.shape)                # one row fewer
```

### 4.3 · Messy text — clean the region labels

```python
print(df_clean["region"].unique())   # notice a stray 'north ' with wrong case/space

df_clean["region"] = df_clean["region"].str.strip().str.title()

print(df_clean["region"].unique())   # now exactly 5 clean labels
```

### ✍️ Practice time — cleaning

## Section 5 · GroupBy & Aggregate

`groupby` splits the table by a column, applies a calculation to each group, and combines the results. This is the single most useful analysis move in pandas.

### 5.1 · Average by region

```python
avg_emp = df_clean.groupby("region")["employment_rate"].mean().sort_values(ascending=False)
print(avg_emp)
```

### 5.2 · Several aggregations at once with `.agg`

```python
summary = df_clean.groupby("region").agg(
    avg_employment=("employment_rate", "mean"),
    total_population=("population", "sum"),
    years=("year", "count"),
).reset_index()
summary
```

### ✍️ Practice time — groupby

Using `df_clean`, find the **average `budget_million` per region**, sorted highest first.

```python
# YOUR CODE HERE
```

**Solution:**

```python
by_budget = df_clean.groupby("region")["budget_million"].mean().sort_values(ascending=False)
print(by_budget)
```

## Section 6 · Merge / Join two tables

Real data is spread across files. `pd.merge` joins them on a shared **key** column. Here we attach each region's **zone** and **employment target** from a lookup table.

```python
lookup = pd.read_csv("session6_region_lookup.csv")
lookup
```

### 6.1 · Left merge on `region`

```python
merged = pd.merge(df_clean, lookup, on="region", how="left")
merged

merged[["region", "year", "employment_rate", "zone", "target_employment"]].head()
```

> **Why `how="left"`?** It keeps every row of the left table (our policy data) and attaches lookup columns where the key matches. If a region were missing from the lookup, those columns would be `NaN` — visible, not silently dropped.

### ✍️ Practice time — merge

```python
# YOUR CODE HERE
```

**Solution:**

## Section 7 · Bringing It Together — a policy analysis

Load → clean → group → merge → insight. The same workflow an analyst runs on any real dataset.

```python
# 1. load
data = pd.read_csv("session6_policy_data.csv")

# 2. clean
data["budget_million"] = data["budget_million"].fillna(data["budget_million"].mean())
data = data.drop_duplicates()
data["region"] = data["region"].str.strip().str.title()

# 3. merge in targets
data = pd.merge(data, pd.read_csv("session6_region_lookup.csv"), on="region", how="left")

# 4. per-region: average employment vs target
report = data.groupby("region").agg(
    avg_employment=("employment_rate", "mean"),
    target=("target_employment", "first"),
).reset_index()
report["meets_target"] = report["avg_employment"] >= report["target"]
report
```

A policymaker can read that table in ten seconds: average employment per region, its target, and a clear yes/no on whether the region is meeting it.
