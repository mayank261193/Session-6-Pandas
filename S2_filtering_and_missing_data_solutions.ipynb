```python
import pandas as pd
df = pd.read_csv("session6_policy_data.csv")
```

### Exercise 1 — A whole row by label  ·  *easy*

Use `loc` to print the first row (index label 0).

```python
print(df.loc[0])
```

### Exercise 2 — One value by position  ·  *easy*

Use `iloc` to print row 0, column 2 (the employment_rate).

```python
print(df.iloc[0, 2])   # 92.3
```

### Exercise 3 — Filter with one condition  ·  *easy*

Keep only rows where `employment_rate` is above 92. How many rows?

```python
high = df[df["employment_rate"] > 92]
print(high.shape[0])
```

### Exercise 4 — Two conditions  ·  *medium*

Keep rows where employment_rate > 92 AND year >= 2027. Each side needs its own ( ).

```python
sel = df[(df["employment_rate"] > 92) & (df["year"] >= 2027)]
sel[["region","year","employment_rate"]]
```

### Exercise 5 — Count the missing  ·  *medium*

Print how many missing values there are in each column.

```python
print(df.isna().sum())   # budget_million shows 2
```

### Exercise 6 — Fill the gaps  ·  *medium*

Fill missing `budget_million` with the column mean. Confirm zero missing after.

```python
df2 = df.copy()
df2["budget_million"] = df2["budget_million"].fillna(df2["budget_million"].mean())
print(df2["budget_million"].isna().sum())   # 0
```

### Exercise 7 — Drop the duplicate  ·  *medium*

Count duplicate rows, then drop them. How many rows remain?

```python
print("dupes:", df.duplicated().sum())   # 1
clean = df.drop_duplicates()
print("rows:", clean.shape[0])   # 50
```

### Exercise 8 — The messy label  ·  *hard*

Print the unique regions — you'll spot a stray one. Then clean them with `.str.strip().str.title()` and print the unique regions again.

```python
print(df["region"].unique())            # note the odd one out
df["region"] = df["region"].str.strip().str.title()
print(df["region"].nunique())           # 5
```

### Exercise 9 — Fix a type  ·  *hard*

Make an integer version of employment_rate (it truncates the decimals). Store it in a new column.

```python
df["emp_whole"] = df["employment_rate"].astype(int)
print(df[["employment_rate","emp_whole"]].head(3))
```

### Exercise 10 — Clean, then filter  ·  *hard*

Region labels are clean now, but the duplicate row is still here. Drop duplicates into `clean_df`, then keep only the North rows and print how many there are.

```python
clean_df = df.drop_duplicates()
north = clean_df[clean_df["region"] == "North"]
print(north.shape[0])   # 10
```
